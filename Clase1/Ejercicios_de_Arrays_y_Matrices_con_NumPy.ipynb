## Ejercicios de Arrays y Matrices con NumPy

```python
## Ejercicio 1: Crear un array de 10 elementos con ceros.
import numpy as np
array_ceros = np.zeros(10)
print(array_ceros)
```

```python
## Ejercicio 1: Crear un array de 10 elementos con ceros.
import numpy as np
array_dos = np.full(10, 0)
print(array_dos)
```

```python
### Ejercicio 2: Genera un array con los números del 1 al 20.
array_num = np.arange(1, 21)
print(array_num)
```

```python
# Ejercicio 3: Crea un array con los números pares del 2 al 50.
array_par = np.arange(2, 51, 2)
print(array_par)
```

```python
# Ejercicio 5: Crea una matriz de 4x4 con valores aleatorios entre 0 y 1.
matriz_aleatoria = np.random.rand(4, 4)
print(matriz_aleatoria)
```

```python
# Ejercicio 6: Realiza la suma elemento a elemento de los arrays a=[1,2,3], b=[4,5,6].
suma_array = np.array ([1,2,3]) + np.array([4,5,6])
print(suma_array)
```

```python
# Ejercicio 7: Multiplica dos matrices 3x3 de enteros aleatorios entre 1 y 10.
matriz_1 = np.random.randint(1, 11, size=(3, 3))
matriz_2  = np.random.randint(1, 11, size=(3, 3))
multiplicacion_matriz = matriz_1 * matriz_2
print(matriz_1)
print(matriz_2)
print(multiplicacion_matriz)
```

```python
# Ejercicio 8: Crea un array de 15 valores aleatorios entre 1 y 100 e imprime su máximo, mínimo y promedio.
array_aleatorio = np.random.randint(1, 101, size=15)
print(array_aleatorio)
print(np.max(array_aleatorio))
print(np.min(array_aleatorio))
print(np.mean(array_aleatorio))
```

```python
# Ejercicio 9: Genera una matriz 5x5 con valores del 1 al 25 y extrae su diagonal.
matriz_5x5 = np.arange(1, 26).reshape(5, 5)
print(matriz_5x5)
# extrae su diagonal
diagonal_matriz = np.diag(matriz_5x5)
print(diagonal_matriz)
```

```python
# Ejercicio 9: Genera una matriz 5x5 con valores del 1 al 25 y extrae su diagonal.
matriz_5 = np.random.randint(1, 11, size=(5, 5))
print(matriz_5)
diagonal_matriz_5 = np.diag(matriz_5)
print(diagonal_matriz_5)
```

```python
# Ejercicio 10 : Dada una matriz 5x5 de enteros aleatorios entre 1 y 20, calcula la suma de cada fila y columna
matriz_aleatoria_1 = np.random.randint(1, 21, size=(5, 5))
print(matriz_aleatoria_1)
# suma de cada fila
suma_filas = np.sum(matriz_aleatoria_1, axis=1)
print(suma_filas)
# Suma de cada columna
suma_columnas = np.sum(matriz_aleatoria_1, axis=0)
print(suma_columnas)
```
