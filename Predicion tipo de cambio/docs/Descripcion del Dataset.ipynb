## 1. Descripción Completa del Dataset

# Resumen General  

 El dataset contiene una serie temporal financiera de la evolución y distribución de créditos o préstamos bancarios hipotecarios y prendarios, segmentados por destino (construcción, refacción, adquisición de automotores, maquinarias, etc.) y por moneda (Pesos Argentinos y Dólares Estadounidenses). 

•	Cantidad de instancias (filas): son 289 instancias en periodos de años y mes aproximadamente 24 años.  
•	Cantidad de características (columnas): 12 variables en total (índices 0 al 11). 

Características y Tipos de Datos (Dtypes) Las columnas se pueden agrupar lógicamente en cuatro categorías principales:

1.	Variables Temporales:  

o	desaño (float64): Representa el año.
o	desmes (object): Representa el mes del registro. 

2.	Operaciones en Moneda Local (Pesos): Todas de tipo numérico entero (int64).

* Préstamos Hipotecarios (H):  

H_Construccion_$,  
H_Refaccion_$,  
H_Adquisición_Nueva_$,  
H_Adquisición_Resto_$,  
H_Adquisición_Otros_$. 

* Préstamos Prendarios/Personales (P):  

P_automotores_$,  
P_maquinarias_$,  
Potros$,  

* Columnas agregadas  

tipo de cambio,  
tasa de interes  
   

3.	Variables Macroeconómicas agregadas en formato (float64):
o	Tipo de cambio,  
o	Tasa de interes,      
La cotización de la moneda y la tasa de interes para el período correspondiente, lo cual se las agregó de otra planilla, el resultado es el promedio del mes, donde es vital para analizar la paridad o convertir montos de USD a Pesos y la tasa de interes para los creditos.

# Proceso de Recopilación y Preprocesamiento
Para llegar a la estructura final de 12 columnas analizadas, se infiere que el dataset pasó por los siguientes procesos:

1.	Extracción y Consolidación: Se extrajeron los reportes mensuales de préstamos otorgados al sector privado y se unificaron en una sola tabla.

2.	Limpieza de Datos:
o	Se transformaron los montos monetarios a números enteros (int64) para facilitar el proyecto. 
o	Se obviaron las columnas  desaño y desmes ya que no son de importancia para el proyecto por el momento. 

3.	Enriquecimiento del Dato: Se añadió la columna Tipo de cambio (float64) y tasa de interes (float64) para cada mes reportado. Estas variables son claves ya que permite homologar y comparar los préstamos otorgados en dólares (_USD) convertidos en pesos y sumados a los otorgados en pesos (_$), estandarizando el valor real del crédito en distintos momentos históricos.
