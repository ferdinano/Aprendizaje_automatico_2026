## 1. Descripción Completa del Dataset

# Resumen General
 El dataset contiene una serie temporal financiera de la evolución y distribución de créditos o préstamos bancarios hipotecarios y prendarios, segmentados por destino (construcción, refacción, adquisición de automotores, maquinarias, etc.) y por moneda (Pesos Argentinos y Dólares Estadounidenses). 

•	Cantidad de instancias (filas): son 289 instancias en periodos de años y mes aproximadamente 24 años.
•	Cantidad de características (columnas): 19 variables en total (índices 0 al 18). 

Características y Tipos de Datos (Dtypes) Las columnas se pueden agrupar lógicamente en cuatro categorías principales:

1.	Variables Temporales:
o	desaño (float64): Representa el año.
o	desmes (object): Representa el mes del registro. 

2.	Operaciones en Moneda Local (Pesos): Todas de tipo numérico entero (int64). 
o	Préstamos Hipotecarios (H): H_Construccion_$, H_Refaccion_$, H_Adquisición_Nueva_$, H_Adquisición_Resto_$, H_Adquisición_Otros_$. 
o	Préstamos Prendarios/Personales (P): P_automotores_$, P_maquinarias_$, Potros$. 

3.	Operaciones en Moneda Extranjera (Dólares - USD): Todas de tipo numérico entero (int64). 
o	Préstamos Hipotecarios (H): H_Construcción_usd, H_Refacción_USD, H_Adquisición_Nueva_USD, H_Adquisición_Resto_USD, H_Adquisición_Otros_USD. 
o	Préstamos Prendarios/Personales (P): P_Automotores_USD, P_Maquinarias_USD, P_Otros_USD. 

4.	Variables Macroeconómicas:
o	Tipo de cambio (float64): La cotización de la moneda para el período correspondiente, lo cual se agregó de otra planilla, el resultado es el promedio del mes, donde es vital para analizar la paridad o convertir montos de USD a Pesos.

# Proceso de Recopilación y Preprocesamiento
Para llegar a la estructura final de 19 columnas analizadas, se infiere que el dataset pasó por los siguientes procesos:

1.	Extracción y Consolidación: Se extrajeron los reportes mensuales de préstamos otorgados al sector privado y se unificaron en una sola tabla.

2.	Limpieza de Datos:
o	Se transformaron los montos monetarios a números enteros (int64) para facilitar el proyecto. 
o	Se validó la existencia de la columna de tiempo dividiéndola en desaño y desmes para permitir agrupar los datos temporalmente (estacionalidad mensual o variaciones anuales). 

3.	Enriquecimiento del Dato: Se añadió la columna Tipo de cambio (float64) para cada mes reportado. Esta variable es clave ya que permite homologar y comparar los préstamos otorgados en dólares (_USD) frente a los otorgados en pesos (_$), estandarizando el valor real del crédito en distintos momentos históricos.
