## Proyecto de Aprendizaje Automático 
# Predicción del Tipo de Cambio y tasa de interés 
A partir del Comportamiento del Crédito Privado (Hipotecario y Prendario). Datos obtenidos de banco Central de la República Argentina, se descargaron dos planillas una de los préstamos y la otra el tipo de cambio y Tasa de interés, para convertirla en una sola planilla.

# 1. Descripción Detallada del Proyecto   

Este proyecto de aprendizaje automático, se centra en predecir  el otorgamiento de créditos al sector privado, las fluctuaciones de la moneda extranjera y la tasa de interés. Utilizando un conjunto de datos que detalla la evolución mensual y anual de los préstamos hipotecarios y prendarios (desglosados por destino como construcción, refacción, adquisición, automotores y maquinarias, tanto en pesos como en dólares), el modelo o los modelos buscarán patrones ocultos que conecten la liquidez del mercado crediticio con la cotización del dólar y la tasa de interés.  

# 2. Formulación de Objetivos  

Objetivo 1:   

• Desarrollar y evaluar un modelo de aprendizaje automático capaz de predecir el "Tipo de cambio" y “tasa de interés” mensual basándose en los volúmenes y “tipos de préstamos hipotecarios y prendarios otorgados en el mercado” a partir del tipo de cambio y tasa de interés.   

Objetivo 2:  

• Identificar el monto de las líneas de financiamiento (por ejemplo, préstamos para maquinarias vs. adquisición de automotores) que presentan un mayor poder predictivo sobre los movimientos del tipo de cambio. 
Métodos de trabajo 
• Comparar el rendimiento de al menos tres métodos de aprendizaje automático para determinar cuál se adapta mejor a la naturaleza volátil de los datos financieros. 
• Generar una herramienta analítica que permita ilustrar de forma práctica y basada en datos las dinámicas macroeconómicas locales y el impacto de las políticas crediticias.  

# 3. Contexto del Problema y su Relevancia.  

El mercado financiero argentino presenta una dinámica macroeconómica muy particular, donde el tipo de cambio y la tasa de interés actúan como un termómetro de las expectativas y la liquidez general. Históricamente, el acceso al crédito a largo plazo (como los hipotecarios) y a mediano plazo (prendarios para bienes de capital o consumo) fluctúa fuertemente en función del escenario cambiario. La relevancia de este problema radica en su doble utilidad:  

• Analítica y Financiera: Permite observar si una expansión o contracción en líneas de crédito específicas (como el freno en préstamos hipotecarios o el aumento en préstamos para la construcción en pesos) actúa como un indicador temprano (o rezagado) de saltos devaluatorios.   

• Económica: Abordar este problema con datos reales permite materializar conceptos macroeconómicos complejos. Como, la deuda privada, la escasez de divisas y el comportamiento del consumidor.  

# 4. Definición del Tipo de Problema.  

El problema abordado se define como un Problema de Regresión. Esto se debe a que las variables objetivas (Target) que el modelo intentará predecir es el "Tipo de cambio" y “Tasa de Interés”, las cuales son variables cuantitativas continuas, y no una categoría discreta o etiqueta.  

# 5. Modelos que se podrían utilizar Para capturar las diferentes naturalezas matemáticas de los datos económicos, se implementarán y compararán los siguientes modelos de regresión:   

1. Support Vector Regresion (SVR): dadas las variables macroeconómicas que no se mueven de manera lineal respecto del volumen de créditos por inflación, devaluaciones abruptas y estacionalidad el SRV, captura dinámicas altamente complejas y no lineales del mercado.  

2. Árbol de Decisión: Dado que las variables macroeconómicas suelen sufrir cambios abruptos, este modelo es altamente relevante. Podrá segmentar los datos creando reglas basadas en umbrales específicos de volumen de préstamos.  

3. K-Means: Este modelo basado en instancias buscará similitudes históricas. Evaluará rangos particulares, como por ejemplo Escenarios estables, Escenarios Volátiles y Escenarios de Crisis.
